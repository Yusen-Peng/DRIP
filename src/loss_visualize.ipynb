```python
import re
import plotly.graph_objects as go


def parse_losses(log_path):
    pattern = re.compile(
        r"cls_loss:\s*([0-9]*\.?[0-9]+)\s*,\s*boundary_loss:\s*([0-9]*\.?[0-9]+)"
    )

    cls_losses = []
    boundary_losses = []

    with open(log_path, "r", encoding="utf-8", errors="ignore") as f:
        for line in f:
            line = re.sub(r"\x1b\[[0-9;]*[A-Za-z]", "", line)
            match = pattern.search(line)
            if match:
                cls_losses.append(float(match.group(1)))
                boundary_losses.append(float(match.group(2)))

    return cls_losses, boundary_losses


def plot_losses_interactive(cls_losses, boundary_losses):
    steps = list(range(len(cls_losses)))

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=steps,
            y=cls_losses,
            mode="lines",
            name="cls_loss",
            yaxis="y1",
        )
    )

    fig.add_trace(
        go.Scatter(
            x=steps,
            y=boundary_losses,
            mode="lines",
            name="boundary_loss",
            yaxis="y1",
        )
    )

    fig.update_layout(
        title="Training Losses",
        xaxis_title="Step",
        # yaxis=dict(title="cls_loss"),
        yaxis=dict(title="boundary_loss", overlaying="y", side="right"),
        hovermode="x unified",
        height=600,
    )

    fig.show()
```

```python
log_file = "../March26_DRIP_temp_01.txt"
```

```python
!pip install nbformat
```

```python
cls_losses, boundary_losses = parse_losses(log_file)
print(f"Parsed {len(cls_losses)} points")
plot_losses_interactive(cls_losses, boundary_losses)
```
